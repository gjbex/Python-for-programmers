# Cut a rope

When you take a rope of length 1, and cut it randomly into three pieces, what is the probability that the pieces can be used as sides of a triangle?

Generating the pieces can be done by selecting two random values, ensuring that they are sorted, and computing the lengths of the pieces.

```python
def cut_rope() -> tuple[float, float, float]:
    cuts = (random.random(), random.random())
    cuts = cuts if cuts[0] < cuts[1] else (cuts[1], cuts[0])
    l1 = cuts[0]
    l2 = cuts[1] - l1
    l3 = 1.0 - l1 - l2
    return l1, l2, l3
```

The condition on the sides of a triangle is that the length of no sides exceeds the sum of the lengths of the two other sides.

```python
def is_triagle(l1: float, l2: float, l3: float) -> bool:
    return l1 <= l2 + l3 and l2 <= l1 + l3 and l3 <= l1 + l2
```

Now you can count the number of cuts that produce the sides of a triangle.

```python
nr_triangles = 0
nr_tries = 100_000
for _ in range(nr_tries):
    l1, l2, l3 = cut_rope()
    if is_triagle(l1, l2, l3):
        nr_triangles += 1
print(nr_triangles/nr_tries)
```

# Center in triangle

Randomly pick three points on a circle with radius 1.  What is the probability that the center of the circle is in the triangle formed by those three points?

It is convenient to define a few helper functions.  The first one compute the cross product between two vectors.  The sign of that product will be useful later on.

```python
def cross_product(p1: tuple[float, float], p2: tuple[float, float]) -> float:
    return p1[0]*p2[1] - p1[1]*p2[0]
```

A second function subtracts two vectors.

```python
def subtract(p1: tuple[float, float], p2: tuple[float, float]) -> tuple[float, float]:
    return (p2[0] - p1[0], p2[1] - p2[1])
```

A function to pick three points on a circle randomly.

```python
def pick_points() -> list[tuple[float, float]]:
    points = []
    for _ in range(3):
        angle = 2*math.pi*random.random()
        points.append((math.cos(angle), math.sin(angle)))
    return points
```

A function that checks whether the center is in the triangle by computing the vector product of a vector representing each point, and a vector from a point to the center of the circle.  If the vector products all have the same sign, the center is in the triangle.

```python
def contains_center(ps: list[tuple[float, float]]) -> bool:
    cps = [cross_product(
               subtract(ps[i], ps[(i + 1) % 3]),
               ps[i]
           ) for i in range(3)]
    return all(map(lambda x: x > 0.0, cps)) or all(map(lambda x: x < 0.0, cps))
```

```python
nr_center_in_triangles = 0
nr_tries = 100_000
for _ in range(nr_tries):
    points = pick_points()
    if contains_center(points):
        nr_center_in_triangles += 1
print(nr_center_in_triangles/nr_tries)
```
